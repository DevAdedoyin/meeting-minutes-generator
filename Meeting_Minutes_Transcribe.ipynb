# **Creating Minutes from a Recorded Meeting**

The goal of this project is to generate meeting minutes from a recorded meeting audio. While this approach can be used to transcribe any audio recording of your choice, I’ll be using a portion of a Denver City Council meeting recording for this project.

The full audio recordings can be found in the [MeetingBank dataset on Hugging Face](https://huggingface.co/datasets/huuuyeah/meetingbank), and the audio files can be downloaded from the [MeetingBank Audio repository](https://huggingface.co/datasets/huuuyeah/MeetingBank_Audio/tree/main).

```python
# importing the necessary libraries

import os
import requests
from IPython.display import Markdown, display, update_display
from openai import OpenAI
from google.colab import drive
from huggingface_hub import login
from google.colab import userdata
from transformers import AutoTokenizer, AutoModelForCausalLM, TextStreamer, BitsAndBytesConfig
import torch
from transformers import pipeline
```

```python
# Constants

LLAMA = "meta-llama/Llama-3.2-3B-Instruct"
```

```python
# Attaching your google drive to colab

drive.mount("/content/drive")
audio_filename = "/content/drive/MyDrive/llms/denver_extract.mp3"
```

```python
# Sign in to HuggingFace Hub

hf_token = userdata.get("HF_TOKEN")
login(hf_token, add_to_git_credential=True)
```

```python
# Open the file
audio_file = open(audio_filename, "rb")
```

# **Step 1: Transcribe Audio**


## **Option 1: Use Open Source for Transcription - Hugging Face Pipelines**

```python
pipe = pipeline(
    "automatic-speech-recognition",
    model="openai/whisper-medium.en",
    dtype=torch.float16,
    device='cuda',
    return_timestamps=True
)

result = pipe(audio_filename)
transcription = result["text"]
display(Markdown(transcription))
```

## **Option 2: Use OpenAI for Transcription**

```python
# Sign in to OpenAI

AUDIO_MODEL = "gpt-4o-mini-transcribe"

openai_api_key = userdata.get("OPENAI_API_KEY")
openai = OpenAI(api_key=openai_api_key)
transcription = openai.audio.transcriptions.create(model=AUDIO_MODEL, file=audio_file, response_format="text")
display(Markdown(transcription))
```

# **Step 2: Analyze and Report**

```python
# System prompt and user prompt

system_prompt = """
You produce minutes of meetings from transcripts, with summary, key discussion points,
takeaways and action items with owners, in markdown format without code blocks.
"""

user_prompt = f"""
Below is an extract transcript of a Denver council meeting.
Please write minutes in markdown without code blocks, including:
- a summary with attendees, location and date
- discussion points
- takeaways
- action items with owners

Transcription:
{transcription}
"""

messages = [
    {"role": "system", "content": system_prompt},
    {"role": "user", "content": user_prompt}
  ]
```

```python
# Quantization configuration instance

quant_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_compute_dtype=torch.bfloat16,
    bnb_4bit_quant_type="nf4"
)
```

```python
# Generate the minute

tokenizer = AutoTokenizer.from_pretrained(LLAMA)
tokenizer.pad_token = tokenizer.eos_token
inputs = tokenizer.apply_chat_template(messages, return_tensors="pt").to("cuda")
streamer = TextStreamer(tokenizer)
model = AutoModelForCausalLM.from_pretrained(LLAMA, device_map="auto", quantization_config=quant_config)
outputs = model.generate(inputs, max_new_tokens=2000, streamer=streamer)
```

```python
# decode the tokens of the meeting
response = tokenizer.decode(outputs[0])
```

```python
# display the response
display(Markdown(response))
```
